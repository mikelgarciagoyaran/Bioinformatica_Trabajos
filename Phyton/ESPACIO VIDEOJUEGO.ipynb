```python
def planeta1():
    print("Nos encontramos ahora en el planeta 1. Aquí reside el alien MEGATON, necesita derrotarlo para comprobar si contiene la cura.")
    print(f"Tu vida es {vida_jugador1} y tu munición es {municion1}. La vida de MEGATON es {vida_megaton}")
    print("1 | Espada láser   | 8 de daño  | 0 de munición")
    print("2 | Disparo básico | 10 de daño | 1 de munición")
    print("3 | Rayo láser     | 18 de daño | 3 de munición")
    print("4 | Bola de fuego  | 25 de daño | 5 de munición")

def ataque_jugador1():
    global municion1, daño_jugador1
    while True:
        seleccion_ataque = input("Seleccione el ataque (1/2/3/4): ")
        if seleccion_ataque == "1":
            daño_jugador1 = 8
            print(f"Has usado espada láser. Tu munición restante es {municion1}")
            break
        elif seleccion_ataque == "2":
            if municion1 > 0:
                municion1 -= 1
                daño_jugador1 = 10
                print(f"Has usado disparo básico. Tu munición restante es {municion1}")
                break
            else:
                print("No te queda munición, prueba con otro ataque")
                #seleccion_ataque = 0
                #break
        elif seleccion_ataque == "3":
            if municion1 > 2:
                municion1 -= 3
                daño_jugador1 = 18
                print(f"Has usado rayo láser. Tu munición restante es {municion1}")
                break
            else:
                print("No te queda munición, prueba con otro ataque")
                #break
        elif seleccion_ataque == "4":
            if municion1 > 4:
                municion1 -= 5
                daño_jugador1 = 25
                print(f"Has usado bola de fuego. Tu munición restante es {municion1}")
                break
            else:
                print("No te queda munición, prueba con otro ataque")
                #break
        else:
            print("No te he entendido")

def ataque_megaton():
    global daño_megaton
    daño_megaton = random.choice(ataques_megaton)

def combate1():
    global vida_megaton, vida_jugador1
    ataque_jugador1()
    ataque_megaton()
    print(f"Tu ataque hace: {daño_jugador1} de daño | Ataque de MEGATON hace : {daño_megaton} de daño")
    if daño_jugador1 > daño_megaton:
        vida_megaton -= daño_jugador1
        print(f"Tu ataque es mayor, GANAS el duelo. MEGATON recibe {daño_jugador1} de daño. Su vida restante es {vida_megaton}")
    elif daño_megaton > daño_jugador1:
        vida_jugador1 -= daño_megaton
        print(f"El ataque de MEGATON es mayor, PIERDES el duelo. Recibes {daño_megaton} de daño. Tu vida restante es {vida_jugador1}")
    else:
        print("Empate. Nadie recibe daño.")
```

```python
def todo_junto():
    if not comienzo_juego():
        print("Hasta la próxima")
        return
    
    print("Bienvenido Doctor")

    if not preguntar_mision():
        print("Hasta la próxima")
        return
        
    print("De acuerdo")
    mostrar_contexto()

    if not explicar_cura():
        print("Hasta la próxima")
        return

    print("De acuerdo")
    mostrar_explicacion_cura()
    
    if not preguntar_preparado():
        print("Hasta la próxima")
        return
    
    print("¡Comienza la investigación!")
   
    mostrar_equipo()
    preguntar_equipo()
    mostrar_armas()
    preguntar_armas()
    mostrar_ataques()
    preguntar_ataques()
    explicar_combate()
    preguntar_combate()
    preguntar_planeta1()
    planeta1()
    while vida_jugador1 > 0 and vida_megaton > 0:
        combate1()

    if vida_megaton <= 0:
        print("¡Has derrotado a MEGATON!")
        print("Ahora necesitas comprobar si su sangre contiene la cura.")
    else:
        print("Has sido derrotado.")
        return

    planeta2()
    while vida_jugador2 > 0 and vida_gigaton > 0:
        combate2()

    if vida_gigaton <= 0:
        print("¡Has derrotado a GIGATON!")
        print("Ahora necesitas comprobar si su sangre contiene la cura.")
    else:
        print("Has sido derrotado.")
        return

    planeta3()
    while vida_jugador3 > 0 and vida_teraton > 0:
        combate3()

    if vida_teraton <= 0:
        print("¡Has derrotado a TERATON!")
        print("Ahora necesitas comprobar si su sangre contiene la cura.")
    else:
        print("Has sido derrotado.")
        return
        
todo_junto()
```
